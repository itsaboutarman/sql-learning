# Retrieving Data From Multiple Tables

```python
import os
from dotenv import load_dotenv

import pandas as pd
import sqlalchemy
```

```python
load_dotenv()

db_host = os.environ.get("db_host")
db_user = os.environ.get("db_user")
db_password = os.environ.get("db_password")
```

```python
engine = sqlalchemy.create_engine(f"mysql+pymysql://{db_user}:{db_password}@{db_host}:3306/sql_store")
```

```python
pd.read_sql("show tables", con = engine)
```

## Inner Joins
We use `ON` to define a specific condition and say: show the row from the second table next to the row from the first table where that condition holds; otherwise, for each row of the main table, it shows all rows of the second table.

```python
query = """
select *
from orders
join customers
"""

pd.read_sql(query, con = engine)
```

```python
query = """
select *
from orders
join customers on orders.customer_id = customers.customer_id
"""

pd.read_sql(query, con = engine)
```

```python
query = """
select order_id, orders.customer_id, first_name, last_name
from orders
join customers on orders.customer_id = customers.customer_id
"""

pd.read_sql(query, con= engine)
```

we can give an ailias to each table to write cleaner code:

```python
query = """
select order_id, o.customer_id, first_name, last_name
from orders o
join customers c 
    on o.customer_id = c.customer_id
"""

pd.read_sql(query, con= engine)
```

EXERCISE:

## Joining Across Databases
we can choose tables from different database to work with them 

```python
query = """
select oi.order_id, oi.product_id, p.name, oi.quantity, oi.unit_price
from order_items oi
join sql_inventory.products p
	on oi.product_id = p.product_id
"""
# note that we are in "sql_store" database but we add another table from "sql_inventory" database

pd.read_sql(query, con= engine)
```

## Self Joins

```python
query = """
select 
    e.employee_id,
    e.first_name,
    m.first_name as manager
from sql_hr.employees e
join sql_hr.employees m
	on e.reports_to = m.employee_id
"""

pd.read_sql(query, con= engine)
```

## Joining Multiple Tables

```python
query = """
select 
	o.order_id,
    o.order_date,
    c.first_name,
    c.last_name,
    os.name as status
from orders o
join customers c
	on o.customer_id = c.customer_id
join order_statuses os
	on o.status = os.order_status_id
"""

pd.read_sql(query, con= engine)
```

EXERCISE:

```python
query = """
select 
	p.date,
    p.invoice_id,
    p.amount,
    c.name,
    pm.name as payment_method
from sql_invoicing.payments p
join sql_invoicing.payment_methods pm
	on p.payment_method = pm.payment_method_id
join sql_invoicing.clients c
	on p.client_id = c.client_id
"""

pd.read_sql(query, con= engine)
```

## Compound Join Condition

```python
query = """
select *
from order_items oi
join order_item_notes oin
	on oi.order_id = oin.order_Id
    and oi.product_id = oin.product_id
"""

pd.read_sql(query, con= engine)
```

## Outer Joins
`OUTER JOIN` in SQL returns rows even when there is no matching row in the other table. The unmatched columns are filled with NULL

Use cases: when you need all records from one table (or both tables with `FULL OUTER JOIN`), not only the intersecting matches—e.g., finding customers with or without orders.

```python
query = """
select p.product_id, p.name, oi.quantity
from products p
left join order_items oi 
	on p.product_id = oi.product_id
"""

pd.read_sql(query, con= engine)
```

this code is same as above but as a best practice avoid using `RIGHT JOIN`:


```python
query = """
select p.product_id, p.name, oi.quantity
from order_items oi
right join products p 
	on p.product_id = oi.product_id
"""

pd.read_sql(query, con= engine)
```

## Outer Joins Between Multiple Tables

```python
query = """
select 
    o.order_id,
	o.order_date, 
    c.first_name,
    sh.name as shipper,
    os.name as status

from orders o

left join customers c
	on o.customer_id = c.customer_id

left join shippers sh
	on o.shipper_id = sh.shipper_id

join order_statuses os
	on o.status = os.order_status_id
    """

pd.read_sql(query, con= engine)
```

## UNIONS

```python
query = """
select 
	order_id,
    order_date,
    "Active" as status
from orders
where order_date >= "2019-01-01"
union
select 
	order_id,
    order_date,
    "Archived" as status
from orders
where order_date <= "2019-01-01"

"""

pd.read_sql(query, con= engine)
```
