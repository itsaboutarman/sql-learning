# Retrieving Data From a Single Table
we use Pandas to import data from mysql and then execute SQL query

```python
import os
from dotenv import load_dotenv

import pandas as pd
import sqlalchemy
```

```python
load_dotenv()

db_host = os.environ.get("db_host")
db_user = os.environ.get("db_user")
db_password = os.environ.get("db_password")
```

```python
engine = sqlalchemy.create_engine(f"mysql+pymysql://{db_user}:{db_password}@{db_host}:3306/sql_store")
```

### The SELECT Statement

```python
df = pd.read_sql("SELECT * FROM customers", con=engine)
df
```

```python
query = """
select *
from customers
where customer_id > 5 
order by first_name
"""


pd.read_sql(query, con= engine)
```

```python
query = """
select 
    last_name,
    first_name,
    concat(first_name, " ", last_name) as full_name
from customers
"""


pd.read_sql(query, con= engine)
```

### DISTINCT

we use `DISTINCT` key word to select unique values from a table:

```python
query = """
select distinct state 
from customers
"""


pd.read_sql(query, con= engine)
```

EXERCISE:

```python
query = """
select 
    name,
    unit_price,
    unit_price * 1.1 as "new price"
from products
"""


pd.read_sql(query, con= engine)
```

### WHERE Clause

Comparison Operators:
<, >, <=, >=, =, !=

```python
query = """
select *
from customers
where state = "va"
"""


pd.read_sql(query, con= engine)
```

EXERCISE:

```python
query = """
select *
from orders
where order_date >= "2019-01-01"
"""


pd.read_sql(query, con= engine)
```

### AND, OR, NOT Operators

```python
query = """
select *
from customers
where (birth_date > "1990-01-01" and not points > 1000) or (points > 3000)
"""


pd.read_sql(query, con= engine)
```

EXERCIS:

```python
query = """
select *
from order_items
where order_id = 6 and (quantity * unit_price > 30)
"""


pd.read_sql(query, con= engine)
```

### IN Operator

```python
query = """
select *
from customers
where state in ("VA", "FL", "GA")
-- where state = "VA" or state = "FL" or state = "GA" 
"""


pd.read_sql(query, con= engine)
```

EXERCISE:

```python
query = """
select *
from products
where quantity_in_stock in (49, 38, 72)
"""


pd.read_sql(query, con= engine)
```

### BETWEEN Operator

```python
query = """
select *
from customers
where points between 1000 and 3000
-- where points >= 1000 and points <= 3000
"""


pd.read_sql(query, con= engine)
```

EXERCISE:

```python
query = """
select *
from customers
where birth_date between "1990-01-01" and "2000-01-01"
"""


pd.read_sql(query, con= engine)
```

### LIKE Operator

`%` : any number of characters

`_` : single character

```python
query = """
select *
from customers
where last_name like "_b" 
"""


pd.read_sql(query, con= engine)
```

### IS NULL Operator

```python
query = """
select *
from customers
where phone is null 
"""


pd.read_sql(query, con= engine)
```

```python
query = """
select *
from customers
where phone is not null 
"""


pd.read_sql(query, con= engine)
```

### ORDER BY Operator 

```python
query = """
select *
from customers
order by state, first_name 
"""


pd.read_sql(query, con= engine)
```

we use `DESC` keyword to reverse the sort order:

```python
query = """
select *
from customers
order by first_name desc
"""


pd.read_sql(query, con= engine)
```

EXERCISE:

```python
query = """
select *
from order_items
where order_id = 2
order by (quantity * unit_price) desc
"""


pd.read_sql(query, con= engine)
```

### LIMIT Clause

```python
query = """
select *
from customers
limit 3
"""


pd.read_sql(query, con= engine)
```

```python
query = """
select *
from customers
limit 6, 3      -- it means that skip the first 6 records and then pick 3 records 
"""


pd.read_sql(query, con= engine)
```
